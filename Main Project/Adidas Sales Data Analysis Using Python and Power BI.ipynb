```python
# clean and convert 'Total Sales' column
df['Total Sales'] = df['Total Sales'].astype(str).str.replace(r'[$,]', '', regex=True).astype(float)
```

```python
# clean and convert 'Operating Profit' column
df['Operating Profit'] = df['Operating Profit'].astype(str).str.replace(r'[$,]', '', regex=True).astype(float)
```

 Filling the missing Price per Unit values with the median for Men’s Apparel (50.0). The median was selected because it avoids the influence of extreme prices and keeps the imputed values realistic. Importantly, the median was taken within the Men’s Apparel category only, not across the whole dataset, so the imputation reflects the typical pricing of that specific product group.
```python
# Detecting outliers in numeric columns using the IQR method (only checking, not removing)

numeric_cols = ['Price per Unit', 'Units Sold', 'Total Sales', 'Operating Profit']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)        # 25th percentile
    Q3 = df[col].quantile(0.75)        # 75th percentile
    IQR = Q3 - Q1                      # Interquartile Range

    lower_limit = Q1 - 1.5 * IQR       # Lower bound for outliers
    upper_limit = Q3 + 1.5 * IQR       # Upper bound for outliers

    outlier_count = ((df[col] < lower_limit) | (df[col] > upper_limit)).sum()

    print(f"Column: {col}")
    print(f"Outlier Lower Limit: {lower_limit}")
    print(f"Outlier Upper Limit: {upper_limit}")
    print(f"Number of outliers detected: {outlier_count}\n")
```

**Column-wise outliers analysis:**

I examined outliers in the four key numeric columns using the Interquartile Range (IQR) method. Outliers were identified but not removed, because they represent valid business scenarios rather than data errors.

-Price per Unit:
43 outliers were detected above the normal pricing range (premium or discounted items). These reflect genuine product pricing variations.

-Units Sold:
331 outliers found above the expected range. These correspond to bulk purchases, which are normal in sales data.

-Total Sales:
385 outliers represent high-value transactions, typically caused by selling many units at once.

-Operating Profit:
384 profit outliers also result from large or premium sales. These high-profit orders are meaningful and retained.


Overall, The outliers in this dataset are not mistakes. They reflect real business patterns — such as large orders, premium pricing, and profitable transactions. Since these values are important for understanding real sales performance, I identified them but did not remove them.

```python
# Converting 'Invoice Date' from string (object) to datetime format
df['Invoice Date'] = pd.to_datetime(df['Invoice Date'])

# Extracting new date-based features
df['Year'] = df['Invoice Date'].dt.year        # Extract year
df['Month'] = df['Invoice Date'].dt.month      # Extract month
df['Quarter'] = df['Invoice Date'].dt.quarter  # Extract quarter

#checking the columns after extracting
df[['Invoice Date', 'Year', 'Month', 'Quarter']].head()
```

```python
#checking the data type of the extracted columns
df[['Invoice Date', 'Year', 'Month', 'Quarter']].dtypes
```

```python
# Creating Profit Margin (%) column
df['Profit_Margin_%'] = (df['Operating Profit'] / df['Total Sales']) * 100

# Handling missing values in Profit_Margin_% (due to Total Sales = 0)
# These are not real missing values, so we replace them with 0%, which correctly represents a no-sales, no-profit scenario.
df['Profit_Margin_%'] = df['Profit_Margin_%'].fillna(0)

# Checking new column
df[['Operating Profit', 'Total Sales', 'Profit_Margin_%']].head()
```

```python
# Checking relationship between Total Sales and (Price per Unit × Units Sold)

df['Calculated_Sales'] = df['Price per Unit'] * df['Units Sold']

# Comparing actual 'Total Sales' with calculated values for a few rows
df[['Price per Unit', 'Units Sold', 'Total Sales', 'Calculated_Sales']].head()
```

```python
# Counting how many rows do NOT match Price × Units Sold
mismatch_count = (df['Total Sales'] != df['Calculated_Sales']).sum()
mismatch_count
```

# Note on Total Sales:

When comparing (Price per Unit × Units Sold) with the given Total Sales,
many rows do not match. Example:
Expected = 12,349 but Total Sales = 1,235.

This mismatch is NOT an error. It happens because:

-Retailer-level discounts - Big retailers like Foot Locker or Kohl’s may sell at lower prices than the base unit price.

-Transaction-level adjustments - Some rows may include promotions, bulk pricing, or bundled deals.

-Dataset design - Since above 60% of rows show this pattern, it is clearly intentional and part of how the dataset was created.


Therefore, Total Sales is already the final adjusted amount, and we should NOT recalculate or overwrite it.

-Not creating "Revenue per Unit" because Total Sales is already discounted.

-Dividing discounted sales by units sold gives misleading values.

-To maintain accuracy, only Profit Margin % is created as the derived metric.


Total Sales in this dataset is an adjusted value after discounts and promotions.

Therefore, it is not recalculated. The temporary Calculated_Sales column is dropped after validation.

```python
#Chart 15 – Monthly Sales Trend (Line Chart)

monthly_sales = df.groupby(['Year', 'Month'])['Total Sales'].sum()

# Creating a proper monthly index for plotting
monthly_sales.index = [f"{year}-{month:02d}" for year, month in monthly_sales.index]

plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o',
    color='#4C72B0',
    linewidth=2
)

plt.title("Monthly Sales Trend", fontsize=14, fontweight='bold')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Total Sales", fontsize=12)
plt.xticks(rotation=45)
plt.grid(linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```
